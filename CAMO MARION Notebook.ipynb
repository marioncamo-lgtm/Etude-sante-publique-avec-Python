<h3 style="margin: auto; padding: 20px; color: RGB(0,150,250); ">3.2 - Nombre théorique de personne qui pourrait être nourries</h3>
</div>

```python
#Combien mange en moyenne un être humain ? Source => #https://www.fao.org/3/i9553fr/I9553FR.pdf
#Selon la FAO, un être humain consomme en moyenne 2 500 kcal par jour.  2000 pour femme et 2500 pour homme entre deux 2250.
#En multipliant ce chiffre par 365 jours, on obtient une consommation annuelle d'environ 912 500 kcal par personne. 
#En convertissant cela en kilogrammes de nourriture, on peut estimer qu'un être humain consomme environ 300 kg de nourriture par an en moyenne. 
#Cependant, il est important de noter que cette estimation peut varier considérablement en fonction de nombreux facteurs tels que l'âge, le sexe, le niveau d'activité physique, les habitudes alimentaires et les conditions socio-économiques.
```

```python
#On commence par faire une jointure entre le data frame population et Dispo_alimentaire afin d'ajouter dans ce dernier la population
dispo_alimentaire_pop = pd.merge(population[population['Année'] == 2017],dispo_alimentaire,   on='Zone', how='inner')
```

```python
#Affichage du nouveau dataframe
(dispo_alimentaire_pop.head(5))
```

```python
#Création de la colonne dispo_kcal avec calcul des kcal disponibles mondialement
dispo_alimentaire_pop['dispo_kcal'] = (
    dispo_alimentaire_pop['Disponibilité alimentaire (Kcal/personne/jour)']
    * dispo_alimentaire_pop['Population']
    * 365)

dispo_kcal_mondial = dispo_alimentaire_pop['dispo_kcal'].sum()
print("Le total des kcal disponibles mondialement est de :", f"{dispo_kcal_mondial:,.0f}".replace(",", " "))
```

```python
#Calcul du nombre d'humains pouvant être nourris
kcal_par_pays = (
    dispo_alimentaire_pop
    .groupby('Zone', as_index=False)['Disponibilité alimentaire (Kcal/personne/jour)']
    .sum()
)

population_par_pays = dispo_alimentaire_pop[['Zone', 'Population']].drop_duplicates()

df = kcal_par_pays.merge(population_par_pays, on='Zone')


df['dispo_kcal'] = (
    df['Disponibilité alimentaire (Kcal/personne/jour)']
    * df['Population']
    * 365
)

df['nb_humains_nourris'] = df['dispo_kcal'] / 912500

total = df['nb_humains_nourris'].sum()
print("Le nombre total d'humains pouvant être nourris est de :", f"{total:,.0f}".replace(",", " "), "soit 8.37 milliards")

#proportion en pourcentage
proportion_humains_nourris = (total / population_totale) * 100
print(f"Ce qui représente : {proportion_humains_nourris:.2f}% de la population totale en 2017")
```
